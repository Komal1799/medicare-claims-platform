```python
import os

SOURCE_DIR = "/Volumes/healthcare_dev/landing/cms_downloads"
BATCH_DIR = f"{SOURCE_DIR}/batches"
RAW_DIR = "/Volumes/healthcare_dev/landing/raw"

FILE_TYPES = ["beneficiary", "carrier", "inpatient", "outpatient",
              "snf", "hha", "hospice", "dme", "pde"]

def release_batch(year):
    """Copy one year's batches into raw, as if the source just delivered them.
    Never overwrites a batch that was already released."""
    released, skipped = [], []

    for file_type in FILE_TYPES:
        src = f"{BATCH_DIR}/{file_type}/batch_{year}"
        dst = f"{RAW_DIR}/{file_type}/batch_{year}"

        if not os.path.exists(src):
            continue                      # no data for this type and year
        if os.path.exists(dst):
            skipped.append(file_type)     # already delivered, leave it alone
            continue

        dbutils.fs.cp(src, dst, recurse=True)
        released.append(file_type)

    print(f"batch_{year} released: {released}")
    if skipped:
        print(f"batch_{year} already in raw, skipped: {skipped}")
```
